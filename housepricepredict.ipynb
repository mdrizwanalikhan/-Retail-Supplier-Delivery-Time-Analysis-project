```python
# 1. Imports & Data Load
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

data = pd.read_csv("kc_house_data.csv")

# 2. Basic Inspection
print(data.head(3))
print(data.info())

# 3. Data Cleaning (33 bedrooms vala outlier fix)
data = data[data['bedrooms'] < 10]

# 4. Visualizations
plt.style.use('seaborn-v0_8-whitegrid')

# Lat vs Long (Geographic distribution)
sns.jointplot(x='lat', y='long', data=data, height=8, color='crimson')
plt.suptitle('Location Distribution', y=1.02)
plt.show()

# Price vs Total Sqft
plt.figure(figsize=(8, 5))
plt.scatter(data['sqft_living'] + data['sqft_basement'], data['price'], alpha=0.4, c='teal')
plt.title("Price vs Total Sqft (Living + Basement)")
plt.xlabel("Total Square Feet")
plt.ylabel("Price ($)")
sns.despine()
plt.show()

# Bedrooms vs Price
plt.figure(figsize=(8, 5))
sns.boxplot(x='bedrooms', y='price', data=data, palette='Set2')
plt.title("Bedrooms vs Price")
sns.despine()
plt.show()
```
